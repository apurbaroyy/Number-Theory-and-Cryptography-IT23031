## Prime numbers
A prime integer has exactly two positive divisors: 1 and itself.

```python
def is_prime(n):
    if n <= 1:
        return False
    for d in range(2, int(n**0.5) + 1):
        if n % d == 0:
            return False
    return True

print([n for n in range(2, 20) if is_prime(n)])
```

## Greatest common divisor
\(\gcd(a,b)=\gcd(b,a\bmod b)\)

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return abs(a)

print(gcd(48, 18))
```

## Modular arithmetic
For positive \(m\), the remainder \(r=a\bmod m\) satisfies \(0\le r<m\).

```python
print(17 % 5)
print(pow(2, 5, 7))
```

## Euler's totient function
\(\varphi(n)\) counts integers from 1 through n that are coprime to n.

```python
def euler_totient(n):
    if n < 1:
        raise ValueError('n must be positive')
    return sum(1 for k in range(1, n + 1) if gcd(k, n) == 1)

print(euler_totient(9))
```
