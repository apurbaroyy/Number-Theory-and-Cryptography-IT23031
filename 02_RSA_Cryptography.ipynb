This toy example uses tiny primes and is not secure.

## RSA mathematics
Choose primes p and q; set n=pq and φ(n)=(p−1)(q−1). Choose e coprime to φ(n), and compute d as the modular inverse of e modulo φ(n).

```python
from math import gcd

p, q = 61, 53
n = p * q
phi = (p - 1) * (q - 1)
e = 17
assert gcd(e, phi) == 1
d = pow(e, -1, phi)

message = 65
ciphertext = pow(message, e, n)
plaintext = pow(ciphertext, d, n)

print('Public key:', (e, n))
print('Private key:', (d, n))
print('Ciphertext:', ciphertext)
print('Decrypted:', plaintext)
assert plaintext == message
```
